```python
import pandas as pd
```

```python
# Task 1

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

print(df)

print("Average marks:", df["Marks"].mean())
print("Maximum marks:", df["Marks"].max())
print("Minimum marks:", df["Marks"].min())
print("Total students:", len(df))
```

```python
# Task 2
print("Students with marks >= 50:")
print(df[df["Marks"] >= 50])

print("\nStudents with marks < 50:")
print(df[df["Marks"] < 50])
```

```python
# Task 3

df["City"] = ["Karachi", "Lahore", "Karachi", "Islamabad", "Karachi"]

print(df)

print("\nStudents from Karachi:")
print(df[df["City"] == "Karachi"])
```

```python
course_data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Course": ["AI", "Web", "AI", "Web", "AI"],
    "Marks": [80, 70, 60, 90, 100]
}

course_df = pd.DataFrame(course_data)

print(course_df)

average_course_marks = course_df.groupby("Course")["Marks"].mean()

print("\nAverage marks of each course:")
print(average_course_marks)
```

```python
# Task 5

missing_data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [80, None, 70, None, 90]
}

missing_df = pd.DataFrame(missing_data)

print("Original DataFrame:")
print(missing_df)

print("\nNumber of missing values:")
print(missing_df["Marks"].isna().sum())

average_marks = missing_df["Marks"].mean()

missing_df["Marks"] = missing_df["Marks"].fillna(average_marks)

print("\nCleaned DataFrame:")
print(missing_df)
```
